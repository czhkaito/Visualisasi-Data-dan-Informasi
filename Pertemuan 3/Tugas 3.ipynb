## Data yang Digunakan
Dataset berupa dataset simulasi yang berisi data tentang penghasilan berbagai profesi di berbagai lokasi, dengan detail sebagai berikut:

- **Kolom Dataset**:
  - `Usia`: Kelompok usia individu, terdiri dari kategori `20-30`, `30-40`, `40-50`, `50-60`.
  - `Profesi`: Jenis pekerjaan individu, meliputi `Dokter`, `Guru`, `Insinyur`, dan `Artis`.
  - `Penghasilan`: Penghasilan rata-rata individu dalam IDR, dengan distribusi normal dan tambahan beberapa outlier.
  - `Lokasi`: Lokasi tempat tinggal individu, terdiri dari kategori `Urban`, `Suburban`, dan `Rural`.
  - `Pengalaman`: Lama pengalaman kerja individu dalam tahun.

- **Jumlah Data**:
  - 500 data utama + beberapa outlier untuk menampilkan variasi data.


## **Tujuan, User, Fungsi, dan Tone**

### **Tujuan**
- Memberikan wawasan tentang distribusi penghasilan berdasarkan profesi, lokasi, usia, dan pengalaman kerja.
- Menunjukkan korelasi antara pengalaman kerja dan penghasilan.
- Menyoroti dampak outlier pada analisis data.

### **User**
- Peneliti ekonomi yang ingin memahami hubungan antara faktor-faktor demografis dengan penghasilan.
- Pembuat kebijakan yang ingin menentukan kebijakan peningkatan kesejahteraan pekerja.
- Pelaku bisnis yang ingin mengetahui tren penghasilan di berbagai profesi.

### **Fungsi**
- Menyediakan visualisasi interaktif dan informatif untuk membantu pengambilan keputusan berbasis data.
- Mengidentifikasi pola distribusi dan outlier dalam data penghasilan.

### **Tone**
- Informal namun profesional, berfokus pada kejelasan dan efisiensi dalam menyampaikan informasi.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Simulasi dataset yang lebih kompleks
np.random.seed(42)
n = 500
data = pd.DataFrame({
    'Usia': np.random.choice(['20-30', '30-40', '40-50', '50-60'], size=n),
    'Profesi': np.random.choice(['Dokter', 'Guru', 'Insinyur', 'Artis'], size=n),
    'Penghasilan': np.random.normal(15_000_000, 5_000_000, size=n),
    'Lokasi': np.random.choice(['Urban', 'Suburban', 'Rural'], size=n),
    'Pengalaman': np.random.randint(1, 30, size=n)
})

# Menambahkan beberapa outlier
outliers = pd.DataFrame({
    'Usia': ['30-40', '40-50'],
    'Profesi': ['Dokter', 'Artis'],
    'Penghasilan': [50_000_000, 5_000_000],
    'Lokasi': ['Urban', 'Rural'],
    'Pengalaman': [15, 5]
})
data = pd.concat([data, outliers], ignore_index=True)
````

````python
# Heatmap unutk korelasi numerik
plt.figure(figsize=(8, 6))
correlation_matrix = data[['Penghasilan', 'Pengalaman']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Korelasi Antara Penghasilan dan Pengalaman')
plt.tight_layout()
plt.show()
````

````python
# Visualisasi scatterplot dengan ukuran berdasarkan pengalaman
plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='Pengalaman', y='Penghasilan', size='Pengalaman', hue='Profesi', alpha=0.7, sizes=(20, 200))
plt.title('Penghasilan vs Pengalaman Berdasarkan Profesi')
plt.xlabel('Pengalaman (Tahun)')
plt.ylabel('Penghasilan (IDR)')
plt.legend(title='Profesi', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
````
